```python
from langchain.agents import tool
@tool
def multiply(a:int,b:int)->int:#retun type ->int
    '''this tool is for multiplication'''
    return a*b
```

```python
multiply.invoke({"a":10,"b":10})
```

```python
multiply.description
```

```python
multiply.args
```

```python
@tool
def get_word_length(word:str)->int:
    '''this tool is used to find out word length''' #docstring
    return(len(word))
```

```python
print(get_word_length.invoke({"word":"my name is lucifer"}))
print(get_word_length("name of eby's dog"))#since one argument no invoke required
```

```python
get_word_length.description
```

```python
get_word_length.args
```
