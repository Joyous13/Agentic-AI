```python
from langchain_groq import ChatGroq
model=ChatGroq(model="gemma2-9b-it")
model
```

```python
## chaining (combining prompt with model)

chain=prompt|model
chain
```

## Assignment:
 Create a simple assistant that uses any LLM and should be pydantic, when we ask about any product it should give you two information product Name, product details tentative price in USD (integer). use chat Prompt Template.

```python
from langchain_groq import ChatGroq
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel,Field

class Product(BaseModel):
    pname: str = Field(desscription="Product name")
    pdetails: str = Field(description="Product details")
    price: int = Field(description="Price of the product in usd")


model=ChatGroq(model="qwen/qwen3-32b")

output_parser=JsonOutputParser(pydantic_object=Product)

prompt=PromptTemplate(
    template="you are a business analyzer.\n{format_instruction}\n{query}\n",
    input_variables=["query"],
    partial_variables={"format_instruction": output_parser.get_format_instructions()},
)

chain=prompt|model|output_parser

chain.invoke({"query":"Iphone 16"})
```
