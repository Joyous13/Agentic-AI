### Text Embedding
https://python.langchain.com/docs/integrations/text_embedding/

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings=OpenAIEmbeddings(model="text-embedding-3-large")

embeddings
```

```python
text="this ios my text on OPENAI embeddings"
query_result=embeddings.embed_query(text)
query_result
```

```python
len(query_result) #the length remains same even after adding one more sentence that is, 3072
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings=OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
```

```python
text="this is my text on OPENAI embeddings"
query_result=embeddings.embed_query(text)
len(query_result)
```

```python
from langchain_community.document_loaders import TextLoader

docs=TextLoader('speech.txt').load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

final_documents=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50).split_documents(docs)

final_documents
```

```python
final_documents[0].page_content
```

```python
embeddings.embed_query(final_documents[0].page_content)
```
